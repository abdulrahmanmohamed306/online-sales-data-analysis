```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the real dataset
df = pd.read_csv('Online Sales Data.csv')

# Data Cleaning & Feature Engineering
df['Total_Revenue'] = df['Units Sold'] * df['Unit Price']

# Financial & Sales KPIs
total_revenue = df['Total_Revenue'].sum()
total_units = df['Units Sold'].sum()
avg_order_value = df['Total_Revenue'].mean()

print(f"--- Real Dataset Analytics Report ---")
print(f"Total Revenue Generated: ${total_revenue:,.2f}")
print(f"Total Units Sold:        {total_units:,}")
print(f"Average Order Value:     ${avg_order_value:,.2f}")

# Groupings
cat_revenue = df.groupby('Product Category')['Total Revenue'].sum().reset_index().sort_values(by='Total Revenue', ascending=False)
region_sales = df.groupby('Region')['Total Revenue'].sum().reset_index()

# Visualizations
sns.set_theme(style="darkgrid")

# 1. Category Chart
plt.figure(figsize=(10, 5))
ax = sns.barplot(x='Product Category', y='Total Revenue', data=cat_revenue, palette='crest')
plt.title('Total Revenue by Product Category', fontsize=14, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
for p in ax.patches:
    ax.annotate(f"${p.get_height():,.0f}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.savefig('real_category_revenue.png', dpi=300)
plt.show()

# 2. Region Chart
plt.figure(figsize=(10, 5))
ax2 = sns.barplot(x='Region', y='Total Revenue', data=region_sales, palette='viridis')
plt.title('Regional Sales Breakdown', fontsize=14, fontweight='bold')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
for p in ax2.patches:
    ax2.annotate(f"${p.get_height():,.0f}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.savefig('regional_sales_performance.png', dpi=300)
plt.show()

print("\nSuccess! The charts have been saved inside your folder.")
```
